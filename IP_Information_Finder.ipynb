```python
import requests

print("Welcome to the IP Checker Program!")

try:
    
    url = "https://ipwho.is/"
    data = requests.get(url, timeout=5).json()

    if data["success"] == False:
        print("Error:", data.get("message", "Something went wrong."))
    else:
        ip = data["ip"]
        country = data["country"]
        city = data["city"]
        isp = data["connection"]["isp"]

        print("IP:", ip)
        print("Country:", country)
        print("City:", city)
        print("ISP:", isp)

        
        if country == "Iran":
            print("You are in Iran.")
        else:
            print("You are not in Iran.")

        if city == "Tehran":
            print("You are in Tehran.")
        else:
            print("You are not in Tehran.")

except requests.exceptions.RequestException as e:
    print("Network error:", e)
```

```python
import socket

target = input("Enter domain or IP: ")
port = int(input("Enter port number (e.g. 80): "))

sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
sock.settimeout(2)
result = sock.connect_ex((target, port))

if result == 0:
    print(f"Port {port} is OPEN")
else:
    print(f"Port {port} is CLOSED")

sock.close()
```
